#Coffee Shop Tea Inventory Automation

This project simulates a small inventory management system for a coffee shop.

The script:
- extracts tea sales from POS PDF reports
- updates the stock automatically
- detects low or critical inventory levels
- exports the updated stock to CSV

Technologies used:
- Python
- Pandas
- pdfplumber

```python
import pandas as pd
import numpy as np
```

# prototype experiment 

```python
df = pd.read_csv("New Microsoft Excel Worksheet.csv")
```

```python
df
```

```python
def stock_status(row):
    if row['Stoc'] < row['Prag_minim']:
        return "CRITIC"
    elif row['Stoc'] == row['Prag_minim']:
        return "low"
    else:
        return "OK"
```

```python
df["Status"] = df.apply(stock_status,axis = 1)
```

```python
df
```

```python
def color_status(val):
    if val == "CRITIC":
        return "background-color: #ff4d4d"   # roșu
    elif val == "LOW":
        return "background-color: #ffd24d"   # galben
    elif val == "OK":
        return "background-color: #85e085"   # verde

df.style.map(color_status, subset=["Status"])
```

```python
tea_sold = "Ceai verde"
quantity = 1
df.loc[df["Ceaiuri"]==tea_sold,'Stoc'] -= quantity
df["Status"] = df.apply(stock_status, axis=1)

df.style.map(color_status, subset=["Status"])
```

```python
import pdfplumber
```

```python
pdf_path = "Top produse vândute 07-03-2026.pdf"
```

```python
with pdfplumber.open(pdf_path) as pdf:
    page = pdf.pages[2]
    text = page.extract_text()
    print (text[:200])
```

# extragere ceaiului vandut

```python
lines = text.split('\n')
```

```python
lines[:10]
```

```python
tea_sales = []
for line in lines:
    if '100 G Buc' in line:
        parts = line.split()
        
     # găsim poziția unde apare '100'
        idx_100 = parts.index('100')
     # numele produsului = cuvintele dintre 1 și '100'
        name = ' '.join(parts[1:idx_100])
        
    # cantitatea e după 'Buc'
        qty = parts[parts.index("Buc")+1]
        qty=int(float(qty))
        
        tea_sales.append((name,qty))
```

```python
lines = [line.strip() for line in meniu_text.split("\n") if line.strip()]
```

```python
# am vazut ca exista un pattern. inainte de fiecare "16 lei", este numele ceaiului 
# asa ca ce am facut , am facut un for loop in care acesta imi ia indexul unde se afla 16 lei , si-i
#cer sa imi dea ce este in urma acestuia , adica numele ceaiului .
```

```python
tea_names = []
for idx in range(len(lines)):
    if lines[idx] == '16 lei':
        tea_names.append(lines[idx-1])
```

```python
# Am observat ca in tea_sales cuvantul 'tea' este scris cu t iar in tea_stock este scris Tea , si asa nu imi va detecta si va ramane NaN
# asa ca facem sa fie la fel scrise toate produsele si acolo si acolo 
```

```python
tea_stock["Produs"] = tea_stock["Produs"].str.strip().str.lower()
sales_df["Produs"] = sales_df["Produs"].str.strip().str.lower()
```

```python
merged = tea_stock.merge(sales_df, on="Produs",how= 'left')
```

```python
merged["Vandut"] = merged["Vandut"].fillna(0).astype(int)
```

```python
merged["Stoc"] = merged["Stoc"] - merged['Vandut'] * 3
```
